```python
import os
import datetime as dt
import multi_inst_plots as m
import matplotlib.pyplot as plt
from IPython.display import clear_output
```

```python
m.options.show()
```

```python
# Weekly plot options
m.options.spacecraft.value = "L1 (Wind/SOHO)"
m.options.resample.value = 60
m.options.resample_mag.value = 15
m.options.resample_stixgoes.value = 0
m.options.l1_av_erne.value = 60
m.options.l1_av_sep.value = 60

m.options.stix.value = False
m.options.goes.value = True
m.options.Vsw.value = True
m.options.T.value = True
m.options.N.value = True
```

```python
fig_save_path = f"{os.getcwd()}{os.sep}weeklyplots"
```

```python
year = 2024
week = 38
while week != 52:
    print(week)
    m.options.startdate.value = dt.date(year,1,1) + dt.timedelta(weeks=week)
    m.options.enddate.value = dt.date(year,1,7) + dt.timedelta(weeks=week)
    data, metadata = m.load_data()
    fig, axs = m.make_plot()
    fig.savefig(f"{fig_save_path}/l1/{year}/l1_{year}_w{week + 1:02d}", bbox_inches="tight")
    plt.close(fig)
    clear_output(wait=True)
    week += 1
```

```python
del fig
del axs
del data
del metadata
fig.savefig(f"{fig_save_path}/l1/{year}/l1_{year}_w{week + 1:02d}", bbox_inches="tight")
```
